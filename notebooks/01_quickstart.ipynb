## 用合成幂律检查 GP 接口
以下 S=b² 是教学数据，不是论文结果或物理求解器标签。真实数据示范见 examples/train_from_csv.py。

```python
import numpy as np
from puffy.surrogate import Surrogate, metrics
b = np.geomspace(.2, 7, 16)
model = Surrogate.fit(b, b**2)
query = np.geomspace(.22, 6.9, 20)
print(metrics(query**2, model.predict(query)["median"]))
print(model.warnings)
```
